```python
import matplotlib.pyplot as plt
```

```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
colors = ["tab:red", "tab:blue", "tab:green", "tab:orange", "tab:purple"]
best = students.index(max(students))
```

```python
plt.figure(figsize=(9, 5))
bars = plt.barh(languages, students, color=colors)
bars[best].set_edgecolor("black")
bars[best].set_linewidth(2.5)

for i in range(len(students)):
    plt.text(students[i] + 0.5, i, students[i], va="center")

plt.gca().invert_yaxis()
plt.title("Students by Programming Language")
plt.xlabel("Number of Students")
plt.ylabel("Languages")
plt.xlim(0, 40)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
bars2 = plt.bar(languages, students, color=colors)
bars2[best].set_edgecolor("black")
bars2[best].set_linewidth(2.5)

for i in range(len(students)):
    plt.text(i, students[i] + 0.5, students[i], ha="center")

plt.title("Students by Programming Language")
plt.xlabel("Languages")
plt.ylabel("Number of Students")
plt.show()
```
