```python
import matplotlib.pyplot as plt
```

```python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]
```

```python
fig = plt.figure(figsize=(9, 6))
ax = fig.add_axes([0.1, 0.1, 0.7, 0.8])

sc = ax.scatter(study_hours, scores, c=ages, cmap="viridis", marker="o", s=80)
cbar = fig.colorbar(sc, ax=ax)
cbar.set_label("Age")

i = scores.index(max(scores))
ax.annotate("Highest score", xy=(study_hours[i], scores[i]),
            xytext=(study_hours[i] - 1.8, scores[i] - 3),
            arrowprops=dict(arrowstyle="->"))

ax.set_title("Study Hours vs Exam Score")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Exam Score")
ax.grid(True, alpha=0.3)

ax2 = fig.add_axes([0.4, 0.17, 0.2, 0.22])
ax2.bar(range(1, len(scores) + 1), scores)
ax2.set_title("Scores", fontsize=8)
ax2.tick_params(labelsize=6)

plt.show()
```
